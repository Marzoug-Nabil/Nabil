# Partie 1 — Configuration de l'environnement

Avant de commencer à programmer, il est essentiel de configurer un environnement de travail propre et isolé. Cela permet d'éviter les conflits entre les versions des librairies utilisées dans différents projets.

## 1.1 Création d'un environnement avec conda

### Qu'est-ce qu'un environnement virtuel?

Un environnement virtuel est un espace isolé qui contient sa propre version de Python et ses propres librairies. Cela permet de travailler sur plusieurs projets sans que les dépendances de l'un n'interfèrent avec celles de l'autre.

### Windows (Anaconda Prompt)

Ouvrir **Anaconda Prompt** depuis la barre de recherche Windows, puis exécuter :

```bash

# Activer l'environnement

# Vérifier l'installation
python --version
```

## 1.3 conda vs venv — Comparaison

| Critère | conda | venv + pip |
|---|---|---|
| Installation | Inclus dans Anaconda | Inclus dans Python |
| Gestion des dépendances | Automatique (résout les conflits) | Manuelle |
| Packages non-Python (C, Fortran) | Oui | Non |
| Poids | Lourd (~3 Go avec Anaconda) | Léger |
| Fichier d'export | `environment.yml` | `requirements.txt` |
| Recommandé pour | Data science, ML | Projets Python généraux |

**Pour ce cours, nous recommandons conda** car il simplifie l'installation des librairies scientifiques.

````python
# Fonction avec plusieurs retours
def statistiques(notes):
    """Retourne la moyenne, le min, le max et l'étendue."""
    moyenne = sum(notes) / len(notes)
    minimum = min(notes)
    maximum = max(notes)
    etendue = maximum - minimum
    return moyenne, minimum, maximum, etendue

moy, mn, mx, et = statistiques([85, 92, 78, 95, 88])
print(f"Moyenne : {moy:.1f}")
print(f"Min : {mn}, Max : {mx}")
print(f"Étendue : {et}")
````

## 8.3 Récursivité

````python
# Factorielle récursive
def factorielle(n):
    if n <= 1:               # Cas de base
        return 1
    return n * factorielle(n - 1)  # Cas récursif

for i in range(1, 8):
    print(f"{i}! = {factorielle(i)}")
````

````python
# Fibonacci récursif
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

print("Suite de Fibonacci :")
for i in range(10):
    print(f"  F({i}) = {fibonacci(i)}")
````

```python
def trier_selection(liste):
    resultat = liste.copy()
    n = len(resultat)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if resultat[j] < resultat[min_idx]:
                min_idx = j
        resultat[i], resultat[min_idx] = resultat[min_idx], resultat[i]
    return resultat
```

```python
def meilleur_etudiant(classe):
    meilleur = max(classe, key=lambda e: calculer_moyenne_etudiant(e))
    return meilleur["nom"], calculer_moyenne_etudiant(meilleur)

def taux_reussite(classe, seuil=60):
    reussis = [e for e in classe if calculer_moyenne_etudiant(e) >= seuil]
```
