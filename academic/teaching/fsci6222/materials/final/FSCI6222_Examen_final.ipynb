## Rappel : Metriques de classification

Avant de commencer, voici un rappel des principales metriques utilisees pour evaluer un modele de **classification binaire**.

### La matrice de confusion

Toutes les metriques se calculent a partir des quatre quantites de la matrice de confusion :

|  | Predit Negatif (0) | Predit Positif (1) |
|---|---|---|
| **Reel Negatif (0)** | **VN** (Vrai Negatif) | **FP** (Faux Positif) |
| **Reel Positif (1)** | **FN** (Faux Negatif) | **VP** (Vrai Positif) |

- **VP** : le modele predit positif et c'est vrai
- **VN** : le modele predit negatif et c'est vrai
- **FP** : le modele predit positif a tort (fausse alarme)
- **FN** : le modele predit negatif a tort (cas manque)

### Les formules

$$\text{Accuracy} = \frac{VP + VN}{VP + VN + FP + FN}$$

$$\text{Precision} = \frac{VP}{VP + FP}$$

$$\text{Recall (Sensibilite)} = \frac{VP}{VP + FN}$$

$$F_1\text{-score} = 2 \times \frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}}$$

### Quand utiliser chaque metrique ?

| Metrique | Question repondue | Quand l'utiliser |
|---|---|---|
| **Accuracy** | Quelle proportion des predictions est correcte ? | Classes equilibrees, cout d'erreur symetrique |
| **Precision** | Parmi les positifs predits, combien sont reellement positifs ? | Quand les **faux positifs** sont couteux (ex : detection de spam, recommandation produit) |
| **Recall** | Parmi les vrais positifs, combien ont ete detectes ? | Quand les **faux negatifs** sont graves (ex : detection de maladie, securite, fraude) |
| **F1-score** | Compromis entre precision et recall | Classes desequilibrees, quand on veut optimiser les deux a la fois |


### Application au probleme du diabete

Dans le cadre de cet examen, vous etudiez la prediction du diabete, ou :
- **VP** = patiente diabetique correctement detectee
- **FN** = patiente diabetique non detectee (cas grave : retard de diagnostic)
- **FP** = patiente saine alarmee a tort (cas moins grave : simple test de confirmation)
- **VN** = patiente saine correctement identifiee

### Comment extraire les coefficients d'un Pipeline

Votre modele est encapsule dans un `Pipeline` qui contient deux etapes : `'scaler'` (StandardScaler) et `'logreg'` (LogisticRegression). Pour acceder a l'objet de regression logistique a l'interieur du pipeline, on utilise l'attribut `named_steps` :

```python
# Acceder au modele de regression logistique dans le pipeline
logreg = pipe_lr.named_steps['logreg']
```

`named_steps` est un dictionnaire qui associe chaque nom d'etape (defini lors de la creation du Pipeline) a l'objet correspondant. Ici, `pipe_lr.named_steps['logreg']` retourne l'objet `LogisticRegression` entraine.

### Les attributs utiles de la regression logistique

Une fois le modele extrait, vous pouvez acceder a ses attributs appris pendant l'entrainement :

- **`logreg.coef_`** : tableau des coefficients $\beta_i$ de chaque variable. Pour une classification binaire, c'est un tableau de forme `(1, n_features)`, donc on prend `logreg.coef_[0]` pour avoir un vecteur 1D.
- **`logreg.intercept_`** : ordonnee a l'origine $\beta_0$ (le biais).
- **`X.columns`** : les noms des variables dans l'ordre original.

### Interpretation des coefficients

Pour une regression logistique :
- Un **coefficient positif** signifie que la variable **augmente** la probabilite d'etre diabetique
- Un **coefficient negatif** signifie que la variable **diminue** cette probabilite
- La **valeur absolue** du coefficient mesure l'**influence** de la variable (plus elle est grande, plus la variable est determinante)

Comme les donnees ont ete standardisees par le `StandardScaler` dans le pipeline, les coefficients sont **directement comparables entre eux** : leur magnitude reflete l'importance relative de chaque variable.
